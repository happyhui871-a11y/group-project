```python
import pandas as pd
import yfinance as yf
from sklearn.linear_model import LinearRegression

# Download Bank of China historical data
# Bank of China ticker on Hong Kong Exchange is '3988.HK'
df = yf.download("3988.HK", start="2024-01-01", end="2024-02-01")

# Reset index so Date is a column
df = df.reset_index()

# Keep only Date and Open columns
df = df[["Date", "Open"]]

# Sort by date
df = df.sort_values(by="Date", ascending=True).reset_index(drop=True)

# Create prediction dates (example: next 3 trading days)
prediction_data = pd.DataFrame({
    "Date": pd.to_datetime(["2024-02-02", "2024-02-05", "2024-02-06"])
})

# Fit linear regression model
model = LinearRegression()
model.fit(df.index.values.reshape(-1, 1), df["Open"])

# Predict open values for the next 3 days
prediction_data["Open"] = model.predict(prediction_data.index.values.reshape(-1, 1))

print(prediction_data)
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from statsmodels.tsa.arima.model import ARIMA

# Download Bank of China historical data
# Ticker: 3988.HK (Bank of China listed in Hong Kong)
df = yf.download("3988.HK", start="2024-01-01", end="2024-02-01")

# Reset index so Date is a column
df = df.reset_index()

# Keep only Date and Open columns
df = df[["Date", "Open"]]

# Set Date as index
df.set_index("Date", inplace=True)

# Ensure Open is float
df["Open"] = df["Open"].astype(float)

# Fit ARIMA model (simple example: AR(1))
model = ARIMA(df["Open"], order=(1, 0, 0))
model_fit = model.fit()

# Forecast next 3 trading days
forecast = model_fit.forecast(steps=3)

# Generate next 3 dates after the last available
next_dates = pd.date_range(start=df.index[-1], periods=4, freq="B")[1:]  # 'B' = business days

# Build prediction DataFrame
prediction_data = pd.DataFrame({
    "Date": next_dates,
    "Open": forecast
})

print(prediction_data)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import yfinance as yf

# Download Bank of China historical data
df = yf.download("3988.HK", start="2024-01-01", end="2024-02-01")

# Reset index so Date is a column
df = df.reset_index()

# Keep only Date, Open, and Volume columns
df = df[["Date", "Open", "Volume"]]

# Sort by date
df = df.sort_values(by="Date", ascending=True).reset_index(drop=True)

# Split the data into training and testing sets
train_size = int(len(df) * 0.8)
train_data = df[:train_size]
test_data = df[train_size:]

# Fit the linear regression model for Volume using Open as predictor
model_volume = LinearRegression()
model_volume.fit(train_data[["Open"]], train_data["Volume"])

# Predict the next Volume based on the latest Open value
next_open = float(df["Open"].iloc[-1])  # ensure it's a scalar float
next_volume = model_volume.predict([[next_open]])  # 2D array: (1 sample, 1 feature)

print("Predicted Open:", next_open)
print("Predicted Volume:", next_volume[0])
```

```python
import pandas as pd
import yfinance as yf
from statsmodels.tsa.arima.model import ARIMA

# Download Bank of China historical data (3988.HK)
df = yf.download("3988.HK", start="2024-01-01", end="2026-01-31")

# Reset index and keep Date, Open
df = df.reset_index()[["Date", "Open"]]

# Sort by date
df = df.sort_values(by="Date", ascending=True).reset_index(drop=True)

# Set Date as index
df.set_index("Date", inplace=True)

# Ensure Open is float
df["Open"] = df["Open"].astype(float)

# Fit ARIMA model (simple AR(1) example)
model = ARIMA(df["Open"], order=(1, 1, 0))  # (p,d,q) tuned for trend
model_fit = model.fit()

# Forecast next 40 business days (covers Feb–Mar 2026)
forecast_steps = 40
forecast = model_fit.forecast(steps=forecast_steps)

# Generate future business dates
next_dates = pd.date_range(start=df.index[-1], periods=forecast_steps+1, freq="B")[1:]

# Build prediction DataFrame
prediction_data = pd.DataFrame({
    "Date": next_dates,
    "Predicted_Open": forecast
})

print(prediction_data.head(10))   # show first 10 predictions
print(prediction_data.tail(10))   # show last 10 predictions (into March 2026)
```
